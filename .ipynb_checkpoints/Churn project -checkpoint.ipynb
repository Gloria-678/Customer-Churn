```python
def get_model_metrics(y_true, y_predict, model_name):
    acc = accuracy_score(y_true, y_predict)
    # We focus precision and recall specifically on the 'Churn' class
    prec = precision_score(y_true, y_predict, pos_label=pos_label_value)
    rec = recall_score(y_true, y_predict, pos_label=pos_label_value)
    cm = confusion_matrix(y_true, y_predict)
    tn, fp, fn, tp = cm.ravel()
    
    return {
        "Model": model_name,
        "Accuracy": acc,
        "Precision (Churn)": prec,
        "Recall (Churn)": rec,
        "False Negatives (Missed)": fn,
        "True Positives (Caught)": tp
    }
```
